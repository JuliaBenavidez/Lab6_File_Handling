```python
#CELL 1 – Student information and folder setup
from pathlib import Path
import csv
import json
import re
import shutil
from collections import Counter

LAST_NAME = "BENAVIDEZ"
STUDENT_ID = "TUPM-26-1835"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(x) for x in STUDENT_ID if x.isdigit())

BASE_DIR = Path("Lab6_File_Handling_Additional_Exercises")
DATA_DIR = BASE_DIR / "data"
OUTPUT_DIR = BASE_DIR / "output"
BACKUP_DIR = BASE_DIR / "backup"

for folder in [DATA_DIR, OUTPUT_DIR, BACKUP_DIR]:
    folder.mkdir(parents=True, exist_ok=True)

print("Student:", LAST_NAME)
print("Student ID:", STUDENT_ID)
print("Seed Digit:", SEED_DIGIT)
print("ID Sum:", ID_SUM)
print("Data folder exists:", DATA_DIR.exists())
print("Output folder exists:", OUTPUT_DIR.exists())
print("Backup folder exists:", BACKUP_DIR.exists())

from pathlib import Path
import csv
import json
import re
import shutil
from collections import Counter

LAST_NAME = "BENAVIDEZ"
STUDENT_ID = "TUPM-26-1835"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(x) for x in STUDENT_ID if x.isdigit())

BASE_DIR = Path("Lab6_File_Handling_Additional_Exercises")
DATA_DIR = BASE_DIR / "data"
OUTPUT_DIR = BASE_DIR / "output"
BACKUP_DIR = BASE_DIR / "backup"

for folder in [DATA_DIR, OUTPUT_DIR, BACKUP_DIR]:
    folder.mkdir(parents=True, exist_ok=True)

print("Student:", LAST_NAME)
print("Student ID:", STUDENT_ID)
print("Seed Digit:", SEED_DIGIT)
print("ID Sum:", ID_SUM)
print("Data folder exists:", DATA_DIR.exists())
print("Output folder exists:", OUTPUT_DIR.exists())
print("Backup folder exists:", BACKUP_DIR.exists())

#CELL 2 – Helper functions
def get_column(headers, choices):
    def clean(text):
        return re.sub(r"[^a-z0-9]", "", text.lower())

    available = {clean(h): h for h in headers}

    for choice in choices:
        if clean(choice) in available:
            return available[clean(choice)]

    raise ValueError(f"Missing column: {choices}")


def save_json(data, filename):
    path = OUTPUT_DIR / filename

    with open(path, "w", encoding="utf-8") as file:
        json.dump(data, file, indent=4)

    return path


def backup_file(source, filename):
    destination = BACKUP_DIR / filename
    shutil.copy2(source, destination)

    print("Backup created:", destination)
    print("Backup exists:", destination.exists())

    return destination


def safe_float(value):
    try:
        return float(value)
    except (ValueError, TypeError):
        return None


#ACTIVITY 1 – WEB SERVER ACCESS LOG ANALYSIS
dataset_path = DATA_DIR / "weblog.csv"

RECORD_LIMIT = 1000 + ID_SUM
TARGET_CLASS = (SEED_DIGIT % 4) + 2

selected_records = []
status_counts = Counter()
url_counts = Counter()

records_processed = 0
invalid_records = 0

with open(dataset_path, "r", encoding="utf-8-sig", newline="") as file:
    reader = csv.DictReader(file)

    ip_col = get_column(reader.fieldnames, ["IP"])
    time_col = get_column(reader.fieldnames, ["Time", "Timestamp"])
    url_col = get_column(reader.fieldnames, ["URL", "Requested URL"])
    status_col = get_column(reader.fieldnames, ["Status", "HTTP Status"])

    for row in reader:
        if records_processed >= RECORD_LIMIT:
            break

        records_processed += 1

        try:
            ip = row[ip_col].strip()
            time = row[time_col].strip()
            url = row[url_col].strip()

            match = re.search(r"\b([1-5]\d\d)\b", row[status_col])

            if not ip or not time or not url or not match:
                invalid_records += 1
                continue

            status = int(match.group(1))
            status_class = status // 100

            status_counts[status] += 1

            if status_class == TARGET_CLASS:
                selected_records.append(row)
                url_counts[url] += 1

        except (AttributeError, TypeError):
            invalid_records += 1

print("Records processed:", records_processed)
print("Target status class:", f"{TARGET_CLASS}xx")
print("Records in target class:", len(selected_records))
print("Invalid records:", invalid_records)
print("Count per status code:", dict(status_counts))
print("Five most requested URLs:", url_counts.most_common(5))

#
filtered_csv = OUTPUT_DIR / f"{LAST_NAME}_web_filtered.csv"
summary_json = OUTPUT_DIR / f"{LAST_NAME}_web_summary.json"

with open(filtered_csv, "w", encoding="utf-8", newline="") as file:
    if selected_records:
        writer = csv.DictWriter(file, fieldnames=selected_records[0].keys())
        writer.writeheader()
        writer.writerows(selected_records)

summary = {
    "student": LAST_NAME,
    "student_id": STUDENT_ID,
    "records_processed": records_processed,
    "target_status_class": f"{TARGET_CLASS}xx",
    "records_in_target_class": len(selected_records),
    "count_per_status_code": dict(status_counts),
    "five_most_requested_urls": url_counts.most_common(5),
    "invalid_records": invalid_records,
    "filtered_csv": filtered_csv.name
}
```
